## Exercise 1: String Formatting with Capitalization

**Objective:** Create a function that formats a string by trimming whitespace and capitalizing the first letter of each word.

**Concepts:** String trimming, splitting, list comprehension, string joining, capitalization

```python
def format_string(text):
    """
    Format a string by trimming whitespace and capitalizing the first letter of each word.
    
    Args:
        text (str): The input string to format
        
    Returns:
        str: Formatted string with proper capitalization
    """
    # Remove leading and trailing whitespace
    text = text.strip()
    
    # Split the string into words
    words = text.split()
    
    # Capitalize the first letter of each word using list comprehension
    words = [word.capitalize() for word in words]
    
    # Join the words back together with spaces
    formatted_text = " ".join(words)
    
    return formatted_text

# Test cases
print(format_string("hello world "))
print(format_string("  python programming  "))
```

## Exercise 2: Advanced Palindrome Detection

**Objective:** Detect palindromes while ignoring non-alphanumeric characters and case differences.

**Concepts:** List comprehension, character filtering, two-pointer technique, case-insensitive comparison

```python
def is_palindrome(text):
    """
    Check if a string is a palindrome, ignoring non-alphanumeric characters and case.
    
    Args:
        text (str): The string to check
        
    Returns:
        bool: True if the string is a palindrome, False otherwise
    """
    # Filter only alphanumeric characters using list comprehension
    filtered_chars = [ch for ch in text if ch.isalnum()]
    
    # Use two-pointer technique to compare characters
    left, right = 0, len(filtered_chars) - 1
    
    while left < right:
        # Compare characters in a case-insensitive manner
        if filtered_chars[left].casefold() != filtered_chars[right].casefold():
            return False
        left += 1
        right -= 1
    
    return True

# Test cases
print(is_palindrome("A man, a plan, a canal: Panama"))
print(is_palindrome("hello"))
print(is_palindrome("race car"))
```

## Exercise 3: Snake Case to Camel Case Conversion

**Objective:** Convert snake_case variable names to camelCase format.

**Concepts:** String splitting, list comprehension, filtering empty strings, string joining

```python
def snake_to_camel(text):
    """
    Convert snake_case to camelCase.
    
    Args:
        text (str): A snake_case string
        
    Returns:
        str: The string converted to camelCase
    """
    # Split the string by underscores
    parts = text.split('_')
    
    # Filter out empty strings (from consecutive underscores)
    parts = [part for part in parts if part]
    
    # Capitalize each part except the first one
    camel_case = parts[0] + ''.join(part.capitalize() for part in parts[1:])
    
    return camel_case

# Test cases
print(snake_to_camel("first_name"))
print(snake_to_camel("first_name__and__last_name"))
print(snake_to_camel("my_variable_name"))
```

## Exercise 4: Word Frequency Analysis

**Objective:** Count word frequencies in a text and return them sorted by frequency (and alphabetically as a tiebreaker).

**Concepts:** String processing, dictionaries, sorting, lambda functions, tuples

```python
def word_frequencies(text):
    """
    Count word frequencies in a text and return sorted results.
    
    Args:
        text (str): The input text to analyze
        
    Returns:
        list: List of tuples (word, frequency) sorted by frequency (descending) and word (ascending)
    """
    # Convert to lowercase for consistent counting
    text = text.lower()
    
    # Clean non-alphanumeric characters by replacing them with spaces
    cleaned = ""
    for ch in text:
        if ch.isalnum():
            cleaned += ch
        else:
            cleaned += " "
    
    # Split into words
    tokens = cleaned.split()
    
    # Count word frequencies using a dictionary
    word_counts = {}
    for token in tokens:
        word_counts[token] = word_counts.get(token, 0) + 1
    
    # Sort by frequency (descending) and then alphabetically (ascending)
    sorted_frequencies = sorted(word_counts.items(), key=lambda item: (-item[1], item[0]))
    
    return sorted_frequencies

# Test case
result = word_frequencies("To be, or not to be.")
for word, count in result:
    print(f"'{word}': {count}")
```

## Exercise 5: Run-Length Encoding and Decoding

**Objective:** Implement run-length encoding (compress consecutive characters) and decoding (decompress).

**Concepts:** String iteration, character comparison, string concatenation, encoding/decoding algorithms

```python
def rle_encode(text):
    """
    Encode a string using run-length encoding.
    Consecutive identical characters are replaced with the character followed by its count.
    
    Args:
        text (str): The string to encode
        
    Returns:
        str: The run-length encoded string
    """
    if not text:
        return ""
    
    encoded = ""
    count = 1
    
    # Iterate through the string starting from the second character
    for i in range(1, len(text)):
        if text[i] == text[i - 1]:
            # Same character, increment count
            count += 1
        else:
            # Different character, append previous char and its count
            encoded += text[i - 1] + str(count)
            count = 1
    
    # Don't forget the last character sequence
    encoded += text[-1] + str(count)
    return encoded

def rle_decode(encoded_text):
    """
    Decode a run-length encoded string back to its original form.
    
    Args:
        encoded_text (str): The run-length encoded string
        
    Returns:
        str: The decoded string
    """
    decoded = ""
    current_char = ""
    current_count = ""
    
    for ch in encoded_text:
        if ch.isdigit():
            # Accumulate digits for the count
            current_count += ch
        else:
            # Process previous character if exists
            if current_char:
                decoded += current_char * int(current_count)
            # Start new character
            current_char = ch
            current_count = ""
    
    # Process the last character
    decoded += current_char * int(current_count)
    return decoded

# Test cases
sample = "aaabbccccaaa"
encoded = rle_encode(sample)
decoded = rle_decode(encoded)

print(f"Original:  {sample}")
print(f"Encoded:   {encoded}")
print(f"Decoded:   {decoded}")
print(f"Match:     {sample == decoded}")
```
