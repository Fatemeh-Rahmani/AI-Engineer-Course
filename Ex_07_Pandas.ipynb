## Setup: Import Libraries

```python
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

## Exercise 1: Load and Explore Dataset

**Objective:** Load a CSV dataset from a URL and display basic information about it.

**Concepts:** Data loading, shape inspection, basic exploration

```python
# Load hotel bookings dataset from GitHub
url = "https://raw.githubusercontent.com/mpolinowski/hotel-booking-dataset/master/datasets/hotel_bookings.csv"
df = pd.read_csv(url)

print("=" * 50)
print("DATASET OVERVIEW")
print("=" * 50)
print(f"Shape: {df.shape[0]} rows × {df.shape[1]} columns")
print(f"\nColumn names:")
print(df.columns.tolist())
```

## Exercise 2: Create Synthetic Data

**Objective:** Generate synthetic customer data with realistic information.

**Concepts:** Random data generation, list comprehension, DataFrame creation, seed for reproducibility

```python
# Generate synthetic customer data
n = 90000

# Lists of common Persian names
first_names = ["Ali", "Sara", "Reza", "Mina", "Omid", "Niloofar", 
               "Hassan", "Fatemeh", "Kian", "Ladan"]
last_names = ["Ahmadi", "Karimi", "Hosseini", "Rahimi", "Jafari", 
              "Moradi", "Ebrahimi", "Shirazi", "Ghasemi", "Noori"]

# Set seed for reproducibility
np.random.seed(42)

# Create synthetic data
new_data = pd.DataFrame({
    "first_name": np.random.choice(first_names, n),
    "last_name": np.random.choice(last_names, n),
    "email": [f"user{i}@example.com" for i in range(n)],
    "phone": [f"+98{np.random.randint(100,999)}{np.random.randint(1000000,9999999)}" 
              for _ in range(n)],
    "credit_card": [f"{np.random.randint(1000,9999)}-{np.random.randint(1000,9999)}-"
                   f"{np.random.randint(1000,9999)}-{np.random.randint(1000,9999)}" 
                   for _ in range(n)]
})

print(f"Synthetic dataset created with {n:,} rows")
print(f"\nFirst 5 rows:")
print(new_data.head())
```

## Exercise 3: Combine Datasets

**Objective:** Merge the original hotel bookings data with the synthetic customer data.

**Concepts:** DataFrame concatenation, axis parameter, data alignment

```python
# Combine original dataset with synthetic customer data (side-by-side)
df_combined = pd.concat([df.reset_index(drop=True), new_data], axis=1)

print("=" * 50)
print("COMBINED DATASET")
print("=" * 50)
print(f"Combined shape: {df_combined.shape[0]:,} rows × {df_combined.shape[1]} columns")
print(f"\nFirst few columns: {df_combined.columns[:5].tolist()}")
print(f"Last few columns: {df_combined.columns[-5:].tolist()}")
```

## Exercise 4: Data Quality Assessment

**Objective:** Inspect data types, structure, and display sample records.

**Concepts:** DataFrame info, head/tail methods, data type inspection

```python
# Display dataset information
print("=" * 50)
print("DATA TYPE AND STRUCTURE")
print("=" * 50)
print(df_combined.info())

print("\n" + "=" * 50)
print("SAMPLE RECORDS (First 5 rows)")
print("=" * 50)
print(df_combined.head())
```

## Exercise 5: Handle Missing Data

**Objective:** Identify and report columns with missing values.

**Concepts:** Missing value detection, value_counts, sorting, filtering

```python
# Check for missing values
missing_values = df_combined.isnull().sum()
missing_sorted = missing_values[missing_values > 0].sort_values(ascending=False)

print("=" * 50)
print("MISSING VALUES ANALYSIS")
print("=" * 50)
if len(missing_sorted) == 0:
    print("No missing values found!")
else:
    print(f"Columns with missing data ({len(missing_sorted)} total):")
    print(missing_sorted)
```

## Exercise 6: Data Cleaning

**Objective:** Remove unnecessary columns to clean up the dataset.

**Concepts:** Column dropping, inplace parameter, data cleaning

```python
# Remove unnecessary columns
if 'company' in df_combined.columns:
    df_combined.drop(columns=['company'], inplace=True)
    print("Column 'company' has been removed from the dataset.")
else:
    print("Column 'company' not found in dataset.")
```

## Exercise 7: Exploratory Data Analysis

**Objective:** Perform statistical analysis on key columns like country, ADR (Average Daily Rate), and length of stay.

**Concepts:** value_counts, max/min operations, filtering, aggregation, descriptive statistics

```python
# 7a) Top 5 countries by booking count
print("=" * 50)
print("7. EXPLORATORY DATA ANALYSIS")
print("=" * 50)
print("\n7a) Top 5 countries by booking count:")
top_countries = df_combined['country'].value_counts().head(5)
print(top_countries)
```

```python
# 7b) Highest and average ADR (Average Daily Rate)
print("\n7b) ADR (Average Daily Rate) Analysis:")
max_adr = df_combined['adr'].max()
mean_adr = round(df_combined['adr'].mean(), 2)
print(f"  Maximum ADR: ${max_adr:.2f}")
print(f"  Mean ADR:    ${mean_adr}")

# Find booking with highest ADR
max_adr_row = df_combined[df_combined['adr'] == max_adr]
print(f"\n  Booking details for highest ADR:")
print(f"  Country: {max_adr_row['country'].values[0]}")
print(f"  Hotel Type: {max_adr_row['hotel'].values[0] if 'hotel' in max_adr_row.columns else 'N/A'}")
```

```python
# 7c) Average length of stay
print("\n7c) Length of Stay Analysis:")
if 'stays_in_weekend_nights' in df_combined.columns and 'stays_in_week_nights' in df_combined.columns:
    total_nights = df_combined['stays_in_weekend_nights'] + df_combined['stays_in_week_nights']
    mean_nights = round(total_nights.mean(), 2)
    print(f"  Mean total nights stayed: {mean_nights} nights")
else:
    print("  Columns for length of stay not found.")
```

```python
# 7d) 5 special requests 
print("\n7d) special requests  Analysis:")

if "total_of_special_requests" in df_combined.columns:
    special_5 = df_combined.loc[df["total_of_special_requests"] == 5, ["first_name", "last_name", "email"]].drop_duplicates()
    print("\n Individuals with 5 special requests (name, last name, email):")
    if len(special_5) == 0:
        print("No records found with total_of_special_requests == 5.")
    else:
        print(special_5.head(20).to_string(index=False))  # show first 20 for brevity
else:
    print("\n Column 'total_of_special_requests' not found in dataset.")
```

## Exercise 8: Advanced Analysis

**Objective:** Perform grouping, filtering, and aggregation operations on the dataset.

**Concepts:** groupby, dropna, apply, lambda functions, aggregation, filtering

```python
# 8) Group names by last name
print("\n" + "=" * 50)
print("8. ADVANCED ANALYSIS")
print("=" * 50)
print("\n8a) Top 5 last names and associated first names:")

top_last_names = df_combined["last_name"].value_counts().head(5).index

name_groups = (
    df_combined[df_combined["last_name"].isin(top_last_names)]
    [["last_name", "first_name"]]
    .drop_duplicates()
    .groupby("last_name")["first_name"]
    .apply(lambda names: sorted(names.unique()))
)

for last_name, first_names in name_groups.items():
    print(f"  {last_name}: {', '.join(first_names)}")
```

```python
# 8b) Guests with most children and babies
print("\n8b) Guests with the maximum number of children and babies:")

# Handle missing values
df_combined['children'] = df_combined['children'].fillna(0)
df_combined['babies'] = df_combined['babies'].fillna(0)
df_combined['total_kids'] = df_combined['children'] + df_combined['babies']

max_kids = int(df_combined['total_kids'].max())
top_kid_guests = df_combined[df_combined['total_kids'] == max_kids][
    ['first_name', 'last_name', 'email', 'children', 'babies', 'total_kids']
].drop_duplicates()

print(f"  Maximum total kids in a booking: {max_kids}")
print(f"  Number of guests with this maximum: {len(top_kid_guests)}")
if len(top_kid_guests) > 0:
    print(f"\n  Sample guests:")
    print(top_kid_guests.head(3).to_string(index=False))
```

```python
# 8c) Top 3 reservation codes
print("\n8c) Most frequent reservation attributes:")

def find_top_attribute(df, col_name, title):
    """Helper function to find top 3 values in a column"""
    if col_name in df.columns and df[col_name].notna().any():
        top3 = (
            df[col_name]
            .dropna()
            .astype(str)
            .replace("", np.nan)
            .dropna()
            .value_counts()
            .head(3)
        )
        if len(top3) > 0:
            print(f"\n  {title}:")
            for i, (value, count) in enumerate(top3.items(), 1):
                print(f"    {i}. {value}: {count} bookings")
            return True
    return False

# Try different columns
columns_to_check = [
    ('agent', 'Top 3 booking agents'),
    ('reservation_status', 'Top 3 reservation statuses'),
    ('assigned_room_type', 'Top 3 assigned room types'),
    ('meal', 'Top 3 meal types')
]

for col, title in columns_to_check:
    if find_top_attribute(df_combined, col, title):
        break
```
