## Exercise 1: Line Plot - Temperature Trends

**Objective:** Create a line plot showing temperature changes throughout the year.

**Concepts:** Line plots, markers, labels, annotations, data visualization

```python
# Monthly temperature data  
months = ["January", "February", "March", "April", "May", "June",
          "July", "August", "September", "October", "November", "December"]

temperatures = [12, 16, 22, 28, 33, 31, 26, 20, 15, 10, 7, 9]

# Create the line plot
plt.figure(figsize=(10, 6))
plt.plot(months, temperatures, marker='o', markersize=8, linestyle='-', 
         color='#1f77b4', linewidth=2)

# Add labels and title
plt.title("Air Temperature Growth Throughout the Year", fontsize=14, fontweight='bold')
plt.xlabel("Months of the Year", fontsize=12)
plt.ylabel("Air Temperature (°C)", fontsize=12)

# Add value labels on each point
for i, temp in enumerate(temperatures):
    plt.text(i, temp + 1.5, str(temp), ha='center', fontsize=9, fontweight='bold')

plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Exercise 2: Bar Chart - Student Enrollment Comparison

**Objective:** Create a grouped bar chart comparing enrollment across two programs over multiple years.

**Concepts:** Bar charts, grouped bars, legends, categorical comparison

```python
# Student enrollment data
years = [2021, 2022, 2023, 2024, 2025]  
students_A = [120, 135, 150, 160, 170]
students_B = [100, 115, 130, 140, 155]

# Set up bar positions
x = np.arange(len(years))
width = 0.35

# Create the figure
plt.figure(figsize=(10, 6))

# Create bars for both programs
plt.bar(x - width/2, students_A, width, label="Program A", 
        color="skyblue", edgecolor='black', linewidth=0.5)
plt.bar(x + width/2, students_B, width, label="Program B", 
        color="orange", edgecolor='black', linewidth=0.5)

# Add labels and title
plt.xlabel("Year", fontsize=12, fontweight='bold')
plt.ylabel("Number of Students", fontsize=12, fontweight='bold')
plt.title("Number of Students in Two Programs Over Five Consecutive Years", fontsize=14, fontweight='bold')
plt.xticks(x, years)
plt.legend(fontsize=11)
plt.grid(True, alpha=0.3, axis='y')
plt.tight_layout()
plt.show()
```

## Exercise 3: Pie Chart - Market Share Distribution

**Objective:** Create a pie chart showing the distribution of sales across products.

**Concepts:** Pie charts, proportions, labels, colors, percentages

```python
# Product sales data
products = ["Product 1", "Product 2", "Product 3", "Product 4", "Product 5"]
sales = [120, 90, 50, 150, 100]

# Create pie chart
plt.figure(figsize=(9, 7))
colors = ["#ff9999", "#66b3ff", "#99ff99", "#ffcc99", "#ff99cc"]
explode = (0.05, 0, 0, 0.1, 0)  # Explode the largest segment (Product 4)

plt.pie(sales, labels=products, autopct='%1.1f%%', colors=colors,
        startangle=90, explode=explode, shadow=True)

plt.title("Share of Each Product in Total Sales", fontsize=14, fontweight='bold')
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle
plt.tight_layout()
plt.show()
```
