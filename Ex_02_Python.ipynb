## Exercise 1: Calculate String Length

**Objective:** Learn two methods to calculate the length of a string
- Using the built-in `len()` function
- Using manual iteration with a counter

**Concepts:** String length, iteration, built-in functions

```python
# Method 1: Using the built-in len() function
text = "Hello"
length = len(text)
print(f"The length of '{text}' is: {length}")

# Method 2: Using manual iteration with a counter
text = "Hello"
length = 0
for character in text:
    length += 1
print(f"The length of '{text}' is: {length}")
```

## Exercise 2: Alternating Case Transformation

**Objective:** Transform a string so that characters at even positions are uppercase and characters at odd positions are lowercase.

**Concepts:** String indexing, modulo operator, iteration, string concatenation, character methods

```python
# Approach 1: Manual string concatenation
text = "Hello"
result = text[0].upper() + text[1].lower() + text[2].upper() + text[3].lower() + text[4].upper()
print(f"Result for '{text}': {result}")

# Approach 2: Using a loop with modulo operator (more scalable)
text = "python"
result = ""
for index in range(len(text)):
    if index % 2 == 0:  # Even positions -> uppercase
        result += text[index].upper()
    else:  # Odd positions -> lowercase
        result += text[index].lower()
print(f"Result for '{text}': {result}")
```

## Exercise 3: String Trimming (Remove Whitespace)

**Objective:** Remove leading and trailing whitespace from a string.

**Concepts:** String methods, whitespace handling, string trimming

```python
# Example: Remove leading and trailing whitespace
text = "  Hello  "
trimmed_text = text.strip()
print(f"Original: '{text}'")
print(f"Trimmed: '{trimmed_text}'")
```

## Exercise 4: Character Replacement

**Objective:** Replace specific characters in a string with different characters.

**Concepts:** String methods, string slicing, character manipulation

```python
# Approach 1: Using the replace() method
word = "apple"
modified_word = word.replace("e", "!")
print(f"Original: '{word}'")
print(f"Modified: '{modified_word}'")

# Approach 2: Using string slicing
word = "apple"
modified_word = word[:-1] + "!"
print(f"Original: '{word}'")
print(f"Modified (last char replaced): '{modified_word}'")
```

## Exercise 5: Palindrome Detection

**Objective:** Create a function that detects whether a string is a palindrome (reads the same forwards and backwards).

**Concepts:** Functions, string reversal, string comparison, string methods

```python
def is_palindrome(text):
    """
    Check if a string is a palindrome (reads the same forwards and backwards).
    
    Args:
        text (str): The string to check
        
    Returns:
        bool: True if the string is a palindrome, False otherwise
    """
    # Remove spaces from the text for comparison
    cleaned_text = text.replace(" ", "")
    
    # Create the reversed version using slice notation
    reversed_text = cleaned_text[::-1]
    
    # Display the comparison for educational purposes
    print(f"Original (no spaces): {cleaned_text}")
    print(f"Reversed: {reversed_text}")
    
    # Check if the original and reversed versions are the same
    return cleaned_text == reversed_text

# Test cases
print("Test 1: 'madam'")
print(f"Is palindrome: {is_palindrome('madam')}\n")

print("Test 2: 'nurses run'")
print(f"Is palindrome: {is_palindrome('nurses run')}\n")

print("Test 3: 'hello'")
print(f"Is palindrome: {is_palindrome('hello')}")
```
