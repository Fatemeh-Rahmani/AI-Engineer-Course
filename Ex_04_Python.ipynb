## Exercise 1: Remove Duplicates While Preserving Order

**Objective:** Remove duplicate elements from a list while maintaining the original order of first occurrence.

**Concepts:** Sets, iteration, list building, order preservation

```python
def remove_duplicates_keep_order(lst):
    """
    Remove duplicate elements from a list while preserving the original order.
    
    Args:
        lst (list): The input list that may contain duplicates
        
    Returns:
        list: A list with duplicates removed, maintaining original order
    """
    seen = set()
    result = []
    
    for item in lst:
        # Only add items we haven't seen before
        if item not in seen:
            seen.add(item)
            result.append(item)
    
    return result

# Test cases
print("Test 1:")
data = [4, 1, 2, 3, 1, 3]
print(f"Original: {data}")
print(f"Deduplicated: {remove_duplicates_keep_order(data)}\n")

print("Test 2:")
data2 = [1, 1, 1, 2, 2, 3]
print(f"Original: {data2}")
print(f"Deduplicated: {remove_duplicates_keep_order(data2)}")
```

## Exercise 2: Rotate List Elements

**Objective:** Rotate a list by k positions (right rotation by default).

**Concepts:** List slicing, modulo arithmetic, edge case handling

```python
def rotate(lst, k):
    """
    Rotate a list by k positions.
    Positive k rotates right, negative k rotates left.
    
    Args:
        lst (list): The input list to rotate
        k (int): Number of positions to rotate
        
    Returns:
        list: The rotated list
    """
    n = len(lst)
    
    # Handle empty list
    if n == 0:
        return lst
    
    # Normalize k to be within list length using modulo
    k = k % n
    
    # Rotate right by k positions
    return lst[-k:] + lst[:-k] if k != 0 else lst

# Test cases
print("Test 1: Rotate right by 2")
original = [1, 2, 3, 4, 5]
print(f"Original: {original}")
print(f"Rotated:  {rotate(original, k=2)}\n")

print("Test 2: Rotate left by 1")
print(f"Original: {original}")
print(f"Rotated:  {rotate(original, k=-1)}\n")

print("Test 3: Rotate by more than list length")
print(f"Original: {original}")
print(f"Rotated:  {rotate(original, k=7)}")
```

## Exercise 3: Chunk List into Smaller Sublists

**Objective:** Split a list into smaller lists of a specified size.

**Concepts:** List slicing, range with step, list comprehension

```python
def chunk(lst, chunk_size):
    """
    Split a list into smaller lists (chunks) of a specified size.
    
    Args:
        lst (list): The input list to chunk
        chunk_size (int): The size of each chunk
        
    Returns:
        list: A list of sublists, where each sublist has chunk_size elements
    """
    result = []
    
    # Iterate through the list with step size = chunk_size
    for i in range(0, len(lst), chunk_size):
        result.append(lst[i:i+chunk_size])
    
    return result

# Test cases
print("Test 1: Chunk [1,2,3,4,5] by size 2")
data1 = [1, 2, 3, 4, 5]
print(f"Result: {chunk(data1, chunk_size=2)}\n")

print("Test 2: Chunk [1,2,3,4,5,6,7] by size 3")
data2 = [1, 2, 3, 4, 5, 6, 7]
print(f"Result: {chunk(data2, chunk_size=3)}\n")

print("Test 3: Chunk by size 1")
print(f"Result: {chunk(data1, chunk_size=1)}")
```

## Exercise 4: Flatten Nested Lists

**Objective:** Flatten a nested list structure into a single-level list.

**Concepts:** Recursion, type checking, list extension, nested data structures

```python
def flatten(lst):
    """
    Recursively flatten a nested list structure into a single-level list.
    
    Args:
        lst (list): A list that may contain nested lists at any depth
        
    Returns:
        list: A flattened list containing all elements in order
    """
    result = []
    
    for item in lst:
        # If item is a list, recursively flatten it
        if isinstance(item, list):
            result.extend(flatten(item))
        else:
            # Otherwise, add the item directly
            result.append(item)
    
    return result

# Test cases
print("Test 1: Simple nested list")
nested1 = [1, [2, [3, 4], 5], 6]
print(f"Original: {nested1}")
print(f"Flattened: {flatten(nested1)}\n")

print("Test 2: Deeply nested list")
nested2 = [1, [2, [3, [4, 5]]]]
print(f"Original: {nested2}")
print(f"Flattened: {flatten(nested2)}\n")

print("Test 3: Already flat list")
flat = [1, 2, 3, 4]
print(f"Original: {flat}")
print(f"Flattened: {flatten(flat)}")
```

## Exercise 5: Run-Length Encoding for Lists

**Objective:** Encode a list using run-length encoding (compress consecutive identical elements).

**Concepts:** Iteration, tuple creation, element comparison, compression algorithms

```python
def run_length_encode(lst):
    """
    Encode a list using run-length encoding.
    Consecutive identical elements are represented as (element, count) tuples.
    
    Args:
        lst (list): The list to encode
        
    Returns:
        list: A list of (element, count) tuples
    """
    # Handle empty list
    if not lst:
        return []
    
    result = []
    current_value = lst[0]
    count = 1
    
    # Iterate through the rest of the list
    for item in lst[1:]:
        if item == current_value:
            # Same element, increment count
            count += 1
        else:
            # Different element, save the tuple and reset
            result.append((current_value, count))
            current_value = item
            count = 1
    
    # Don't forget the last sequence
    result.append((current_value, count))
    
    return result

# Test cases
print("Test 1: Basic encoding")
data1 = [1, 1, 1, 2, 2, 3, 3, 3, 3]
result1 = run_length_encode(data1)
print(f"Original: {data1}")
print(f"Encoded:  {result1}\n")

print("Test 2: All same elements")
data2 = [5, 5, 5, 5]
result2 = run_length_encode(data2)
print(f"Original: {data2}")
print(f"Encoded:  {result2}\n")

print("Test 3: No consecutive duplicates")
data3 = [1, 2, 3, 4]
result3 = run_length_encode(data3)
print(f"Original: {data3}")
print(f"Encoded:  {result3}\n")

print("Test 4: Empty list")
data4 = []
result4 = run_length_encode(data4)
print(f"Original: {data4}")
print(f"Encoded:  {result4}")
```
