## Exercise 1: Create and Inspect 3D Arrays

**Objective:** Create a 3D random array and inspect its shape and properties.

**Concepts:** NumPy array creation, random number generation, shape attribute, dimensionality

```python
import numpy as np

# Create a 3D random array (4x5x3)
array_3d = np.random.rand(4, 5, 3)

print("Shape of the 3D array:", array_3d.shape)
print("Total number of elements:", array_3d.size)
print("Number of dimensions:", array_3d.ndim)
print("\nFirst few elements of the array:")
print(array_3d)
```

## Exercise 2: Reshape Arrays

**Objective:** Reshape a 3D array into a 2D array without changing data order.

**Concepts:** Array reshaping, maintaining data, shape transformation

```python
# Create a 3D random array
array_3d = np.random.rand(4, 5, 3)
print("Original 3D array shape:", array_3d.shape)

# Reshape to 2D (20 rows, 3 columns)
# Note: 4 * 5 * 3 = 60 elements → 20 * 3 = 60 elements (maintains data)
array_reshaped = array_3d.reshape(20, 3)
print("Reshaped 2D array shape:", array_reshaped.shape)
print("\nFirst 5 rows of reshaped array:")
print(array_reshaped[:5])
```

## Exercise 3: Array Indexing

**Objective:** Access elements in a 2D array using indexing notation.

**Concepts:** Array indexing, row-column access, zero-based indexing

```python
# Create a 2D array
arr = np.array([[10, 20, 30, 40], 
                [50, 60, 70, 80]])

print("2D array:")
print(arr)
print("\nAccessing element at row 1, column 0:")
print("arr[1][0] =", arr[1][0])
print("arr[1, 0] =", arr[1, 0])  # Alternative indexing notation

print("\nAccessing entire row 0:")
print(arr[0])

print("\nAccessing entire column 2:")
print(arr[:, 2])
```

## Exercise 4: Data Type Conversion

**Objective:** Convert array data types efficiently.

**Concepts:** Type casting, astype() method, data type awareness

```python
# Create an integer array
arr = np.array([[10, 20, 30, 40], 
                [50, 60, 70, 80]])

print("Original array (integer type):")
print(f"Data type: {arr.dtype}")
print(arr)

# Convert to float
arr_float = arr.astype(float)

print("\nConverted array (float type):")
print(f"Data type: {arr_float.dtype}")
print(arr_float)

# Convert to int32
arr_int32 = arr.astype(np.int32)
print(f"\nConverted to int32: {arr_int32.dtype}")
```

## Exercise 5: Concatenate Arrays

**Objective:** Combine multiple arrays along an axis.

**Concepts:** Array concatenation, axis parameter, array combining

```python
# Create two 1D arrays
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

print("Array 1:", arr1)
print("Array 2:", arr2)

# Concatenate along axis 0 (default for 1D arrays)
concatenated = np.concatenate((arr1, arr2))
print("\nConcatenated 1D arrays:", concatenated)

# Example with 2D arrays
arr_2d_1 = np.array([[1, 2], [3, 4]])
arr_2d_2 = np.array([[5, 6], [7, 8]])

print("\n2D Array 1:")
print(arr_2d_1)
print("\n2D Array 2:")
print(arr_2d_2)

# Concatenate along rows (axis=0)
concat_rows = np.concatenate((arr_2d_1, arr_2d_2), axis=0)
print("\nConcatenated along rows (axis=0):")
print(concat_rows)

# Concatenate along columns (axis=1)
concat_cols = np.concatenate((arr_2d_1, arr_2d_2), axis=1)
print("\nConcatenated along columns (axis=1):")
print(concat_cols)
```
