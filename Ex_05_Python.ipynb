## Exercise 1: Create Dictionary from Keys and Values

**Objective:** Create a dictionary from two lists of keys and values, handling mismatched lengths gracefully.

**Concepts:** Dictionary creation, enumeration, zip-like operations, None values

```python
def make_dict(keys, values):
    """
    Create a dictionary from two lists of keys and values.
    If values list is shorter, missing values are set to None.
    Extra values are ignored.
    
    Args:
        keys (list): List of keys
        values (list): List of values
        
    Returns:
        dict: A dictionary mapping keys to values
    """
    result = {}
    
    for i, key in enumerate(keys):
        # Use None if there's no corresponding value
        if i < len(values):
            result[key] = values[i]
        else:
            result[key] = None
    
    return result

# Test cases
print("Test 1: More keys than values")
keys1 = ["a", "b", "c"]
values1 = [1]
print(f"Keys: {keys1}, Values: {values1}")
print(f"Result: {make_dict(keys1, values1)}\n")

print("Test 2: More values than keys")
keys2 = ["a", "b", "c"]
values2 = [1, 3, 4, 5, 6]
print(f"Keys: {keys2}, Values: {values2}")
print(f"Result: {make_dict(keys2, values2)}\n")

print("Test 3: Equal length")
keys3 = ["x", "y", "z"]
values3 = [10, 20, 30]
print(f"Keys: {keys3}, Values: {values3}")
print(f"Result: {make_dict(keys3, values3)}")
```

## Exercise 2: Sort Pairs and Count Elements

**Objective:** Sort pairs by the second element, extract first elements, and count their frequencies.

**Concepts:** Sorting with lambda functions, list comprehension, dictionary counting, tuple unpacking

```python
def count_and_sort_desc(pairs):
    """
    Sort pairs by the second element, extract first elements, and count their frequencies.
    
    Args:
        pairs (list): List of tuples (first_element, second_element)
        
    Returns:
        tuple: (sorted_first_elements, sorted_counts)
               - sorted_first_elements: First elements in order after sorting pairs by second element
               - sorted_counts: Elements and their frequencies sorted by frequency (descending)
    """
    # Sort pairs by the second element (converting to int for numerical sorting)
    sorted_pairs = sorted(pairs, key=lambda x: int(x[1]))
    
    # Extract the first elements from sorted pairs
    first_elements = [x[0] for x in sorted_pairs]
    
    # Count frequencies of first elements
    counts = {}
    for num in first_elements:
        counts[num] = counts.get(num, 0) + 1
    
    # Sort counts by frequency in descending order
    sorted_counts = sorted(counts.items(), key=lambda x: x[1], reverse=True)
    
    return first_elements, sorted_counts

# Test case
pairs = [(1, "4"), (1, "2"), (2, "1"), (3, "3"), (3, "1"), (4, "3"), (3, "3")]
order_list, counts_sorted = count_and_sort_desc(pairs)

print("Original pairs:", pairs)
print("Sorted list (by 2nd element):", order_list)
print("Sorted count (by frequency):", counts_sorted)
```

## Exercise 3: Invert Dictionary Mapping

**Objective:** Convert a dictionary with department as keys and lists of names as values to one with names as keys and department as values.

**Concepts:** Dictionary iteration, nested loops, dictionary building, sorted iteration

```python
def invert_mapping(dept_to_names):
    """
    Invert a department-to-names mapping to names-to-department mapping.
    
    Args:
        dept_to_names (dict): Dictionary with departments as keys and lists of names as values
        
    Returns:
        dict: Dictionary with names as keys and department as value
    """
    name_to_dept = {}
    
    # Iterate through departments and their associated names
    for dept, names in dept_to_names.items():
        # Process names in sorted order for consistency
        for name in sorted(names):
            name_to_dept[name] = dept
    
    return name_to_dept

# Test case
data = {
    "CV": ["sara"],
    "ML": ["ali", "mina"],
    "AI": ["fatemeh", "alex"]
}

result = invert_mapping(data)
print("Original mapping (Dept -> Names):", data)
print("Inverted mapping (Names -> Dept):", result)
```

## Exercise 4: Deep Merge Dictionaries

**Objective:** Merge two dictionaries recursively, combining nested dictionaries instead of replacing them.

**Concepts:** Dictionary merging, recursion, type checking, nested structures

```python
def deep_merge(dict1, dict2):
    """
    Recursively merge two dictionaries.
    When both values are dictionaries, they are merged recursively.
    Otherwise, values from dict2 override values from dict1.
    
    Args:
        dict1 (dict): The first dictionary (base)
        dict2 (dict): The second dictionary (to merge into dict1)
        
    Returns:
        dict: A new merged dictionary
    """
    # Create a copy to avoid modifying the original
    result = dict1.copy()
    
    # Iterate through dict2 items
    for key, value in dict2.items():
        # If the key exists in result and both values are dicts, merge recursively
        if key in result and isinstance(result[key], dict) and isinstance(value, dict):
            result[key] = deep_merge(result[key], value)
        else:
            # Otherwise, override with value from dict2
            result[key] = value
    
    return result

# Test case
print("Test 1: Nested dictionary merge")
A = {"a": {"x": 1, "y": 2}, "b": 0}
B = {"a": {"y": 9, "z": 3}, "c": 7}
merged = deep_merge(A, B)

print(f"Dict A: {A}")
print(f"Dict B: {B}")
print(f"Merged: {merged}\n")

print("Test 2: Deep nesting")
A2 = {"x": {"y": {"z": 1}}}
B2 = {"x": {"y": {"w": 2}}}
merged2 = deep_merge(A2, B2)

print(f"Dict A: {A2}")
print(f"Dict B: {B2}")
print(f"Merged: {merged2}")
```
