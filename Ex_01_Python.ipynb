### Conceptual Questions

1. **Difference between interpreted and compiled languages:**  
   Compiled languages (e.g., C++) translate code to machine code before running. Interpreted languages (e.g., Python) execute code line-by-line at runtime. Compiled are faster but less flexible; interpreted aid quick development.

2. **Why Python for AI?**  
   Simple syntax, vast libraries (TensorFlow, PyTorch), strong community, and easy data handling with NumPy/Pandas make it ideal for AI prototyping and deployment.

3. **Variable naming rules:**  
   Start with letter or underscore; use letters, numbers, underscores after. Case-sensitive; no keywords; no spaces/special chars. Use descriptive names.

4. **Difference between `name` and `Name`:**  
   Case-sensitive, so they are distinct variables. Can hold different values.

5. **Persian names for variables?**  
   Allowed in Python 3 (Unicode support), but not recommended due to readability, compatibility, and convention issues. Use English.
